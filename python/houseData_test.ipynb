```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../Dataset/house_prices.csv')
```

```python
df.describe()
```

```python
df['Carpet Area'] = df['Carpet Area'].combine_first(df['Super Area'])
# df.dropna(subset=['Carpet Area'], inplace=True)
df['Carpet Area'] = (
    df['Carpet Area']
    .str.extract(r'(\d+\.?\d*)')
    .astype(float)
)
df_filled = df.fillna(0)
```

```python
df_select = df_filled.loc[:, ['Index', 'Price (in rupees)', 'Carpet Area']]
```

```python
from rtree_bindings import Rect, RTree
```

```python
df_select
```

```python
t = RTree()
for i in range(df_select.shape[0]):
    rect = Rect(df_select.iloc[i, 1], df_select.iloc[i, 2])
    t.insert(rect, df_select.iloc[i, 0], df_select.iloc[i, 0])
```

```python
res = t.search(Rect(19900, 500, 20000, 700))
for payload, id_ in res:
    # print("payload:", payload, "id:", id_)
    print("rect:", df_select.iloc[id_, 1], ",", df_select.iloc[id_, 2], "id:", id_)

print('--------------')


del_info = t.delete(Rect(20000.0 , 700.0), 2436, 2436)
# del_info = t.delete(Rect(20000.0 , 700.0), 2437, 2437)
if (del_info): 
    print("delete! 2436")
res = t.search(Rect(19900, 500, 20000, 700))
for payload, id_ in res:
    print("rect:", df_select.iloc[id_, 1], ",", df_select.iloc[id_, 2], "id:", id_)
```
